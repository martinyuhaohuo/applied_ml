# Linear Regression

## Formulation of Linear Regression

### Linear Functions Family
- As a prominent hypothesis calss (model space), linear functions family is the space of linear functions $𝑓_𝛽(𝑥)$ defined by the dot product $𝑓_𝛽(x) = 𝛽^⊤x$
- $\beta \in R^d$ is the parameter / weight vector
- $x\in R^d$ is the input vector (also called features or covariates)
- $\hat 𝑦 = 𝑓_𝛽(𝑥)$ is called the (predicted) label (also called output or response or target)
- Different values of $𝛽$ correspond to different functions $𝑓_𝛽(.)$

### Linear Regression Problem
- Input: dataset  $Z = (𝑥_1, 𝑦_1), (𝑥_2, 𝑦_2), . . . ,(𝑥_n, 𝑦_n)$, where $𝑥_𝑖 ∈ ℝ^𝑑$ and $𝑦_𝑖 ∈ ℝ$
- Desired Output: a linear function $𝑓_𝛽(x) = 𝛽^⊤x$ such that $𝑦_𝑖 ≈ 𝛽^⊤𝑥_𝑖$, or the parameter vector $𝛽$ that determines such a function
- Typical notation:
    - Use $𝑖$ to index examples $𝑥_𝑖, 𝑦_𝑖$ in dataset $Z$
    - Use $𝑗$ to index components $𝑥_𝑗$ of $𝑥 ∈ ℝ^d$
    - $𝑥_{𝑖𝑗}$ is component $𝑗$ of the $𝑖$th input example vector
- Goal: 
    - Estimate $𝛽 ∈ ℝ^d$ such that $𝑦_𝑖 ≈ 𝛽^⊤𝑥_𝑖$, the approximately equal means that the linear function fits the labels of the examples well
    - More formally, we need to define a loss function to evaluate how well the function estimated fits the labels of the examples

### Loss Function: Mean Squared Error
- A very common choice of loss function for linear regression problem
- Intuition:
    - For a single example $y_i$, the squared distance between the prediction and the label meqasures the wellness of fit
    - Therefore, the overall wellness of fit of a function to a dataset can be defined as the average of the squared distance between each label and the prediction
- Mean squared error definition: $ℒ(𝛽; 𝑍) = \frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2$
- The mean squared error is computationally convenient and works well in practice; thus it is the standard choice of loss function for regression problems
- The mean squared error is convex in $\beta$ (“bowl shaped”), implying that there exists a global optimum $\beta$ that minimizes the MSE
- Contour plots: each contour corresponds to a set of $\beta$ that leads to a fixed value of $ℒ$

### Linear Regression Algoriothm
- Input: dataset  $Z = (𝑥_1, 𝑦_1), (𝑥_2, 𝑦_2), . . . ,(𝑥_n, 𝑦_n)$, where $𝑥_𝑖 ∈ ℝ^𝑑$ and $𝑦_𝑖 ∈ ℝ$
- Desired Output: a linear function $𝑓_{\hat \beta}(x) = \hat \beta^⊤x$ that minimizes the MSE $ℒ(𝛽; 𝑍) = \frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2$
- In other word, linear regression finds the best-fit function from the linear function space.
- The parameter vector $\hat \beta$ is derived through solving
$$
\begin{align}
    \hat \beta (Z) &= argmin_{\beta \in R^d} ℒ(𝛽; 𝑍) \\
    &= argmin_{\beta \in R^d} \frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2 \\
    &= argmin_{\beta \in R^d} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2
\end{align}
$$
- With these choices, the linear regression problem is sometimes called “Ordinary Least Squares” (OLS)

## Automatic Model Generation as Function Approximation

### Function Approximation View of ML
- Presume there is an unknown function $𝑓^∗$ that generates $Z$ (with some measurement noise)
- Given dataset $Z$, ML algorithm finds an approximation $𝑓_𝛽 ≈ 𝑓^*$ in our “model family” $𝑓_𝛽 ∈ ℱ$
- However, often, $𝑓^∗$ won’t even be realizable within our model family $ℱ$; for example $f^*(𝑥) = x^2$ is not realizable inside in linear functions model family
- Two key design choice questions under this view:
    - What is the family $ℱ$ of candidate models (what model space may include the true nature function)
    - How to define “approximating” $𝑓^∗$ (how to design loss function):
        - Approximating the true data-generation function is achieved through finding the function that best fits the training data
        - However, the training data are limited, often noisy, measurements of the true data-generation function
        - This is why we should deisgn the loss function to make the model fit only “the regularities in the data”

## Linear Regression with Feature Maps

### Feature Map
- Why we need feature map:
    - Linear regression can approximate the true data generation function well if the data generation function belongs to the linear function space.
    - However, what if the data generation function is non-linear? Using linear regression with feature map is one solution as it extends the hypothesis space beyond linear function space
- What is a feature map:
    - A feature map is defined as a function $𝜙: 𝑋 → ℝ^𝑑$, which map the original feature vector $x$ to the remapped/expanded feature vector $𝜙(x)$
- Linear Regression with feature map:
    - We can apply linear regression upon a given non-linear feature map, so that even the hypothesis space is linear in $𝜙(x)$, it is non-linear in $x$, $ℱ = 𝑓_𝛽(𝑥) = 𝛽^⊤𝜙(x)$
    - The MSE is $\frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝜙(x_i))^2$
    - In turn by combing linear regression with feature map, we can conduct non-linear regression

### Types of Feature Map
- Adding intercept term:
    - $𝜙(𝑥) = [1, 𝑥_1, …, 𝑥_𝑑]⊤$, almost always used as the first term captures constant effect
- Polynomial features:
    - $𝝓(𝑥) = [1, 𝑥_1, 𝑥_2, 𝑥_1^2, 𝑥_1𝑥_2, 𝑥_2^2, ...]$
    - Degree 2 polynomial feasture maps are very common as they capture “feature interactions”
    - Can use other nonlinearities (exponential, logarithm, square root)
- Encoding non-real inputs:
    - Suppose $𝑥 ∈ {“high school”, “college”, “masters”, “doctoral”}$, $𝜙(𝑥)$ maps $x$ to {1, 2, 3, 4}$

## Overfitting and Underfitting

### Is Expanding the Hypothesis Space Always Good
- By using more complex feature map, we can expands the number of expanded features in linear regression and lead to more expressive hypothesis set
- Such expressive hypothesis set may allows the linear regression algorithm to select a function that perfectly matches all datapoint (e.g. a n-1 degree polynomial can fit n datapoints perfectly)
- However, such result is not what we need; the goal of machine learning is generalization, we fit model to the datapoints to approximate the true data generatiopn function and capture regularity
- A model that fits all datapoint perfectly fits the noise, rather than learning the regularities in data
- As a result, for new datapoints with unseen noise, the model generates high prediction error

### Definition of Overfitting and Underfitting
- Training data: examples $𝑍 = {(𝑥, 𝑦)}$ used to fit the model
- Test data: new inputs $𝑥$ whose labels $𝑦$ is not seen before by the model
- Overfitting: 
    - Fit the training data 𝑍 well; fit new test data 𝑥, 𝑦 poorly
    - The model fit the noise that are not relevant to the regularity
    - Cause: too many hypotheses in $ℱ$ that all fit the data well (overly expressive hypothesis set relative to data, or too much noise in data)
- Underfitting: 
    - Fit the training data 𝑍 poorly; usually also fit new test data 𝑥, 𝑦 poorly
    - Cause: inexpressive hypothesis space such that it contains no function that can approximate the true data generation function

### Cause of Noise
- Measurement error in the feature values
- Measurement / annotation error in the target labels
- True labels may depend on features not amongst $x_{ij}$
- Errors due to data entry and storage 

## Assessing Overfitting

### Train and Test Split
- ML engineer usually split the whole dataset they have into train and test data, in order to measure generalization and diagnose overfitting/underfitting
- Training/Test Split Protocol:
    - Shuffle and split $𝑍$ into $𝑍_{train}$ and $𝑍_{test}$
    - Run linear regression with $𝑍_{train}$ to obtain $\hat \beta(Z_{train})$
    - Evaluate training loss: $ℒ_{train} = ℒ(\hat \beta (𝑍_{train}) ; 𝑍_{train})$
    - Evaluate testing loss: $ℒ_{test} = ℒ(\hat \beta (𝑍_{test}) ; 𝑍_{test})$, which is an estimate of true generalization loss
- Overfitting: $ℒ_{train}$ is small, $ℒ_{test}$ is large
- Underfitting: $ℒ_{train}$ is large, $ℒ_{test}$ is large

### Why Randomly Sample the Train and Test Set
- We can randomly sample $𝒁_{𝒕𝒓𝒂𝒊𝒏}$ and $𝒁_{𝒕𝒆𝒔𝒕}$ from the same $𝒁$ as long as IID assumption holds
    - That is, when IID assumption holds, the testsing loss should approximate the model's performance on new data well
    - IID assumption: all data are independent and identically distributed from the same data distribution $𝑃(𝑥, 𝑦)$
    - This is a strong (but common) assumption
- For data that evolves over time (data distribution may shift over time), split train and test set differently along time
    - In general, split train/test data based on how we expect the test inputs to differ from train data

### Three ways to fix underfitting and overfitting
- Improve the training dataset (quality and quantity)
- Choose the right model family (expand or reduce hypothesis space)
- Choose the right loss function (use regularization to account both model fitness and complexity)

## Bias and Variance Tradeoff

### Definitions of Bias and Variance
Imagine we draw $𝑘$ training datasets from the same probability distribution over data, and each time fit the model $\{𝑓_{\hat \beta}\}_{1:k} = \{𝑓_{\hat \beta (Z_{train, 1})}, ...,  𝑓_{\hat \beta (Z_{train, k})}\}$ to it
- Variance: how much do those fitted functions $\{𝑓_{\hat \beta}\}_{1:k}$ differ amongst each other, on average
- Bias: how much does the average of all those fitted functions $\{𝑓_{\hat \beta}\}_{1:k}$ deviate from the true function over the data distribution
- Variance, bias, overfitting, and underfitting:
    - In general, the more powerful the function class (hypothesis space), the more the ML algorithm tortures itself to fit all the noise in the training data
    - As the noise pattern changes every time when a new training dataset is drawn, the model fitted will also change significantly, leading to higher variance
    - On the other hand, the complex hypothesis space allows the ML algorithm to finds the true data generation function in average, leading to low bias
    - Therefore, overfitting is characterized by high variance and low bias, while underfitting is characterized by low variance and high bias
    - Ideally, we want to restrict the hypothesis space to the extent that it just contain the true data generatyion function, so that both bias and variandce will be low

### Theoretical Result on MSE
- Generalization MSE ≈ bias^2 + variance, where generalization MSE is the expectation of MSE w.r.t all possible training dataset $Z_{train}$
- Implication:
    - As size of training data increases, variance would continue to diminish, while the bias will not diminish
    - This is because bias is the lowest achievable error corresponding to the “best model” within the hypothesis space
    - While as the dataset size grows, different i.i.d. datasets would induce similar fits, leading to lower variance
- Hence, as dataset size grows, generalization error would be dominated by bias rather than variance:
    - To reduce error, we select high capacity (more complex hypothesis space), low bias models when the size of dataset is large
    - When datasize is small or data quality is not good, we choose models with more restricted hypothesis space, as they are less affected by data quality and quantity

### Case of Linear Regression with Feature Map
- For linear regression with feature maps, increasing feature dimension 𝑑 tends to increase capacity, decrease bias but increase variance
- Need to construct $𝝓$ to balance tradeoff between bias and variance, the rule of thumb is that we will need $𝑛 ≈ 10 𝑑’$ samples

## Regularization

### What is Regularization
- Overfitting happens as in original loss function, we only asked the ML algorithm to fit the training data as well as possible, so it produced overly complex fits
- Regularization as the solution: regularization modifies the loss function so that the ML algorithm not only fit the training data but also try to simplify the complexity of the generated model: $ℒ(𝛽; 𝑍) =$ Train MSE + Fit complexity
- Regularizers: terms in loss function that express other desiderata beyond fitting the training data

### L2 Regularization for Linear Regression
- L2 regularization modifies the loss function by adding a measure of fit complexity (use L2 norm of parameter vector)
$$
\begin{align}
ℒ(𝛽; 𝑍) &= \frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2 + \lambda || 𝛽 ||_2^2 \\
&= \frac{1}{n} \sum_{i = 1}^{n} (y_i - 𝛽^⊤𝑥_𝑖)^2 + \lambda \sum_{j = 1}^{d} 𝛽_j^2 \\
\end{align}
$$
- 𝜆 is a “hyperparameter” that must be tuned (satisfies 𝜆 ≥ 0)
- L2 regularization penalize $|| 𝛽 ||_2^2$, pulls coefficients towards 0; as $𝜆 → ∞$, it forces $𝛽_𝑗 = 0$
- Common convention: 
    - If using intercept term, no penalty on the intercept/bias parameter $𝛽_0$
    - As $𝜆 → ∞$, we have $𝛽_1 = ⋯ = 𝛽_𝑑 = 0$, and thus only fit $𝛽_0$ (which yields $\hat \beta_0(𝑍) = mean(\{𝑦_𝑖\}_{𝑖=1}^𝑛$)

### Intuition of L2 Regularization: Soft-Constraining F
- Consider a feature map with 𝑑 = 10 features $[1,𝜙_1, … ,𝜙_𝑑]$
- Suppose that the data generation function has $𝛽_𝑗 = 0$ for all $𝑗 > 5$
- This means that we should choose a smaller-capacity hypothesis class, induced by the smaller feature map with 𝑑 = 5
- Forcing $𝛽_𝑗$’s to be 0 for all $𝑗 > 5$ by using a feature map with smaller dimension is hard-constraining F
- An alternative way is to use L2 regularization to encourage $𝛽_𝑗$’s to have small magnitude for all $𝑗 > 5$
- This will make the learner focus on the preferred subsets of the hypothesis class F

### Intuition of L2 Regularization: Gaussian Priors over 𝛽
- L2 regularized linear regression amounts to preferring smaller weights according to a Gaussian “prior” pdf
- That is, without ever looking at the data, the algorithm like smaller $𝛽_𝑗$ than large $𝛽_𝑗$ (prior distribution of $\beta$ follows a multi-variable Gaussian distribution centered at 0, so again soft-constraining F)
- So the larger value of $𝛽_𝑗$ is only selected for the model if it is much better for the data fit (MSE)
- Before regularization, the preference of the algorithm over $\beta$ is unformly distributed, after L2 regularization, the preference is Gaussian distributed

### Effect of 𝜆 on Bias and Variance
- With 𝜆 too high, the algorithm finds overly simple model, low variance but high bias
- With 𝜆 too low, the algorithm finds overly complex model, low bias but high variance

### General Regularization Strategy
- $ℒ_new(𝛽; 𝑍) = ℒ(𝛽; 𝑍) + 𝜆 ⋅ 𝑅(𝛽)$, where the regularization $𝑅(𝛽)$ is independent of data
- With regularization $𝑅(𝛽)$, we have used it as a route to inject our preferences for “simpler” functions within our function class $F$
- In otherword, hypothesis class gives all function that the ML algorithm can choose, regulization allows us to inject our preference over function avaliable in that space

## Hyperparameter Tuning

### Tuning 𝜆 on Test Set
- 𝜆 is a hyperparameter that must be tuned (satisfies 𝜆 ≥ 0)
- Assume we know the range for where 𝜆 can be, based on prior experience, we may try a few different candidates $𝜆_𝑡$
and choose the one that minimizes the test set loss (often makes sense to sample uniformly in log-space e.g.10^{−3}, 10^{−2}, 10^{−1}, 1, 10,10^2, 10^3)
- This, however is problematic:
    - Test set MSE, while noisy, is meant to be an unbiased estimate of true generalization MSE
    - Tuning 𝜆 on test set leaks test set info into our model
    - Test set error gets biased to be lower than true generalization error (overly optimistic)
    - This may be thought of as overfitting of hyperparameters on the test set

### Tuning 𝜆 on Validation Set
- Set aside data for hyperparameter selection (validation set), disjoint from data for estimating generalization performance (“test set”) and also from train set
- Basic held-out validation protocol:
    - Split $𝑍$ into $𝑍_{train}$, $𝑍_{val}$, and $𝑍_{test}$
    - For $𝑡 ∈ \{1, … , ℎ\}$ hyperparameter choices:
        - Run linear regression with $𝑍_train$ and $𝜆_𝑡$ to obtain $\hat \beta(𝑍_{train}, 𝜆_t)$
        - Evaluate validation error, $ℒ_{val}^t = ℒ(\hat \beta(𝑍_{train}, 𝜆_t); 𝑍_{val})$
    - Use best $𝜆_t$:
        - Choose $t' = argmin_t ℒ_{val}^t$ with lowest validation error
        - Return $\hat \beta(𝑍_{train}, 𝜆_{t'})$ or use $𝑍_{train} ∪ 𝑍_{val}$ if Z is small

### Tuning 𝜆 with 𝑘-Fold Cross-Validation
- Split $𝑍$ into $𝑍_{train}$ and $𝑍_{test}$
- Split $𝑍_{train}$ into $𝑘$ disjoint validation sets $𝑍_{val}^𝑠$, where $𝑠 ∈ {1, … , 𝑘}$
- For each $𝑠$:
    - Lect the fold $𝑍_{val}^𝑠$ be the validation set
    - Let the training set contain all other folds: $𝑍_{train}^𝑠 = \bigcup_{𝑠' \neq 𝑠} 𝑍_{val}^{𝑠'}$
    - For each candidate $𝜆_𝑡$, train one model on each $𝑍_{train}^𝑠$ and evaluate it on the corresponding $𝑍_{val}^𝑠$
- Compute the average validation loss across $𝑠 ∈ {1, … , 𝑘}$ for each $𝜆_𝑡$, choose $𝜆'$ that performs best
- Retrain linear regression on the whole $𝑍_{train}$ using the selected $𝜆'$, then evaluate once on $𝑍_{test}$
- This strategy generally chooses a better $𝜆'$ than a single validation split
- Compute vs. accuracy tradeoff: as $𝑘 → 𝑛$, model selection becomes more accurate, but the algorithm becomes more computationally expensive

## Performance Metrics

### Alternative Performance Metrics
- Mean absolute error: $\frac{1}{n} \sum_{i = 1}^{n} |\hat y_i - y_i|$
- Mean relative error: $\frac{1}{n} \sum_{i = 1}^{n} \frac{|\hat y_i - y_i|}{|y_i|}$
- $R^2$ score: $1 - \frac{MSE}{stdev^2}$
    - Also called the coefficient of determination
    - $stdev$ is the standard deviation of the label $y$
    - Higher is better; $R^2 = 1$ is perfect
- Pearson correlation between predictions and labels:
    $$
    \frac{1}{n} \sum_{i = 1}^{n} \frac{(\hat y_i - \hat \mu)(y_i - \mu)}{\widehat{stdev} \; stdev}
    $$
    - $\mu$ and $stdev$ are the mean and standard deviation of labels
    - $\hat \mu$ and $\widehat{stdev}$ are the mean and standard deviation of predicted labels
- Rank-order correlation: measures the degree of monotonicity between $\hat y_i$ and $y_i$, rather than the degree of linearity

### Performance Metrics vs. Loss Functions
- Use performance metrics tied closely to the application when tuning hyper-parameter or evaluating model's generalization performance
- Use a performance metric that is convenient for computation as loss functions (should also expresses all the performance metrics we care about)
- When evaluating model for some performance measure, always compare it against some simple baselines like:
    - Randomly predicting some sample from the training label
    - Predicting the mean / median of the training labels
    - Any other hand-coded guess we can construct without much effort
