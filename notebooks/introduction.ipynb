# What is Machine Learning

## Four Views on Machine Learning

### ML as Performance Imporvement by a System
- Machine Learning is the study of algorithms that improve their performance $P$ at some task $T$ with experience or training data $Z$.
- A well-defined learning task is given by $(P, T, Z)$

### ML as Automatic Generation of Models from Data
- ML take data as input and generate a model as output
- A model is a structured representation of knowledge (function) about the world from which that data was sampled
- Models are useful because they make predictions; Given new input $x$, a model $f(.)$ outputs a predicted output $f(x)$

### ML as Information Compression
- Dataset size: $n$ observations and $d$ features and $1$ label imply that the dataset size is $n(d+1)$ floats
- Model size; suppose that the model has $k$ parameters, then the model size is $k$ floats
- The information inputed takes $n(d+1)$ floats to store, while the output is only $k$ floats
- Learning a model is an act of compression: distilling the regularities of the world behind the data into an abstraction that can generalize
- LLMs: it is this similar massive compression that forces the model to discover underlying patterns across the internet, rather than simply memorize it

### ML as Programming by Examples
- Instead of producing a detailed programme to solve a specific problem, why not rather try to produce an algorithm that can learn itself on how to solve problem from examples
- The latter way of "programming" is machine learning
- When to use machine learning over traditional programming: high data quantity and quality, low analytical modelling and understanding
- Examples: use ML for recognizing animals from pictures, use traditioal programming when flying rockets to other planets

## Machine Learning Design Choices

### Hypothesis Class
- The class of functions from which the ML procedure must pick one to fit the data

### Loss Function
- Defines what it means for a function to fit the data well
- A function that is high for bad fits, low for good fits

### Optimizer
- The algorithm to search for the function that best fits the data
